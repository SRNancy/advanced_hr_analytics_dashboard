```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
# KPI values
total_employees = df.shape[0]
employees_left = df[df['Attrition']== 'Yes'].shape[0]

attrition_rate = round((employees_left/total_employees)*100, 2)

avg_age = round(df['Age'].mean(), 0)

avg_income = round(df['MonthlyIncome'].mean(), 0)

# STYLE
sns.set_style('whitegrid')

# CREATE DASHBOARD
fig, axes = plt.subplots(2,2, figsize=(16,10))

fig.suptitle('Advanced HR Analytics Dashboard', fontsize=20, fontweight='bold', y=1.02)



# 1. HEATMAP

numeric_df = df.select_dtypes(include='number')

corr = numeric_df.corr()

sns.heatmap(corr,
            cmap='coolwarm',
            ax=axes[0,0])
axes[0,0].set_title('Correlation Heatmap')

# 2. SALARY VS ATTRITION (BOXPLOT)

sns.boxplot(x='Attrition',
            y='MonthlyIncome',
            data=df,
            ax=axes[0,1])
axes[0,1].set_title('Salary vs Attrition')

# 3. JOB ROLE VS ATTRITION

sns.countplot(x='JobRole',
              hue='Attrition',
              data=df,
              ax=axes[1,0])
axes[1,0].set_title('Job Role vs Attrition')

# 4. AGE DISTRIBUTION

axes[1,1].hist(df['Age'], bins=20)
axes[1,1].set_title('Employee Age Distribution')
axes[1,1].set_xlabel('Age')
axes[1,1].set_ylabel('Count')

# KPI TEXT

plt.figtext(0.08, 0.94, f'Total Employees: {total_employees}', fontsize=11, fontweight='bold')
plt.figtext(0.28, 0.94,f'Employees Left: {employees_left}', fontsize=11, fontweight='bold')
plt.figtext(0.46, 0.94, f'Attrition Rate: {attrition_rate}%', fontsize=11, fontweight='bold')
plt.figtext(0.66, 0.94, f'Average Income: {avg_income}',fontsize=11, fontweight='bold')
plt.figtext(0.86, 0.94, f'Average Age: {avg_age}', fontsize=11, fontweight='bold')

# SPACING
plt.tight_layout(rect=[0,0,1,0.88])

# SAVE DASHBOARD
plt.savefig('advanced_hr_dashboard.png', bbox_inches='tight')

# Plot dash
plt.show()



```
